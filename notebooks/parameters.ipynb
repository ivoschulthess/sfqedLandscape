# Parameter Calculations

```python
import numpy as np
from scipy.constants import physical_constants
```

```python
# speed of light in [m/s]
clight = physical_constants['speed of light in vacuum'][0]

# reduced Planck constant in [eV s]
hbar_eVs = physical_constants['reduced Planck constant in eV s'][0]

# reduced Planck constant in [J s]
hbar_Js = physical_constants['reduced Planck constant'][0]

# electron mass in [MeV]
me_MeV = physical_constants['electron mass energy equivalent in MeV'][0]

# electron mass in [kg]
me_kg = physical_constants['electron mass'][0]

# elementary charge in [C]
Qe_C = physical_constants['elementary charge'][0]

# fine-structure constant
alpha = physical_constants['fine-structure constant'][0]
```

```python
# QED electric field scale (Schwinger field) in [V/m]
E_qed = me_kg**2 * clight**3 / Qe_C / hbar_Js

# QED magnetic field scale in [T]
B_qed = E_qed / clight
```

## Bunch

```python
# bunch charge in [C]
bunchCharge = 3e-9

# number of particles in one bunch
N_bx = bunchCharge / Qe_C


print(f'N:\t{N_bx:.2e}')
```

## Electron-Laser

### Laser Intensity

```python
##########
# INPUT  #
##########

# laser intensity parameter
a0 = 4.1

# laser wavelength in [µm]
laserWavelength = 0.8

# electron energy in [MeV]
E_e = 6000

# collision angle in [deg]
theta = 0


##########
# OUTPUT #
##########

# laser angular frequency in [rad/s]
omega_l = 2*np.pi*clight / (laserWavelength * 1e-6)

# laser photon energy in [eV]
E_l = hbar_eVs * omega_l

# electron Lorentz factor
gamma_0 = E_e / me_MeV

# quantum parameter
chi = a0 * gamma_0 * E_l * (1 + np.cos(theta*np.pi/180)) / (1e6 * me_MeV)

# energy parameter
eta = chi / a0


print(f'a0:\t{a0:.3e}')
print(f'χ:\t{chi:.3e}')
print(f'η:\t{eta:.3e}')
```

### Pulse Energy

```python
##########
# INPUT  #
##########

# laser pulse energy in [J]
laserPulseEnergy = 2

# laser wavelength in [µm]
laserWavelength = 0.8

# laser waist in [µm]
waist = 10

# laser pulse duration fwhm in [fs]
laserPulseDuration = 35

# polarization factor LP: sqrt(2), CP: 1
polarizationFactor = np.sqrt(2)

# electron energy in [MeV]
E_e = 6000

# collision angle in [deg]
theta = 0


##########
# OUTPUT #
##########

# laser angular frequency in [rad/s]
omega_l = 2*np.pi*clight / (laserWavelength * 1e-6)

# laser photon energy in [eV]
E_l = hbar_eVs * omega_l

# electron Lorentz factor
gamma_0 = E_e / me_MeV

# laser intensity parameter
a0 = 147.839 * polarizationFactor * np.sqrt(laserPulseEnergy) * laserWavelength / waist / np.sqrt(laserPulseDuration)

# quantum parameter
chi = a0 * gamma_0 * E_l * (1 + np.cos(theta*np.pi/180)) / (1e6 * me_MeV)

# energy parameter
eta = chi / a0


print(f'a0:\t{a0:.3e}')
print(f'χ:\t{chi:.3e}')
print(f'η:\t{eta:.3e}')
```

```python
# factor for the laser pulse shape
# Gaussian:   0.94
# sech^2:     0.88
# ignore it:  1.00
shapeFactor = 1

# peak power in [W]
peakPower = shapeFactor * laserPulseEnergy / (laserPulseDuration*1e-15)

print(f'peak power: {peakPower/1e12:.1f} TW')
```

## Synchrotron / FEL

```python
##########
# INPUT  #
##########

# magnetic field strength of the undulator in [T]
B_0 = 1.42

# electron energy in [MeV]
E_e = 8000
E_e = 17500

# undulator period in [m]
lambda_u = 68e-3

# undulator length in [m]
L_u = 105


##########
# OUTPUT #
##########

# electron Lorentz factor
gamma_0 = E_e / me_MeV

# phase duration in [rad]
phi = 2 * np.pi * L_u / lambda_u

# undulator strength parameter
# corresponds to the xi / a0
# K ~ 93.4 * B_0 [T] * lambda_u [m]
K = Qe_C * B_0 * lambda_u / 2 / np.pi / me_kg / clight

# quantum parameter
chi = gamma_0 * B_0 / B_qed

# energy parameter
eta = chi / K

# radiation reaction parameter
RR = 2/3 * alpha * chi * K * phi


print(f'K (a0):\t{K:.3e}')
print(f'χ:\t{chi:.3e}')
print(f'η:\t{eta:.3e}')
print(f'RR:\t{RR:.3e}')
```
